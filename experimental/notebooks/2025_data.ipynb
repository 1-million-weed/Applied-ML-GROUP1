```python
import fastf1
import datetime
import pandas as pd
```

```python
year = 2025
schedule = fastf1.get_event_schedule(year)
current_date = datetime.datetime.now().date() #in year-month-day format

def race_finsihed(date, current_date):
    """Check if the event date has passed."""
    return date < current_date

passed_races = []

for event_date  in schedule['EventDate']:
    event_date = pd.Timestamp(event_date).date()
    if race_finsihed(event_date, current_date):
        if schedule[schedule['EventDate'] == event_date.strftime('%Y-%m-%d')]['EventFormat'].values[0] != 'testing':
            passed_races.append(schedule[schedule['EventDate'] == event_date.strftime('%Y-%m-%d')])
    elif not race_finsihed(event_date, current_date):
        event_date = event_date.strftime('%Y-%m-%d')
        upcomming_race = schedule[schedule['EventDate'] == event_date]
        break
        
```

```python
for race in passed_races:
    round_number = race['RoundNumber'].values[0]
    qualifying_session = fastf1.get_session(year, 7, 'Q')
    race_session = fastf1.get_session(year, round_number, 'R')


race_session.load()
race_session.laps
```
