```python
import fastf1
import datetime
import pandas as pd
```

```python
year = 2025
schedule = fastf1.get_event_schedule(year)
current_date = datetime.datetime.now().date() #in year-month-day format

def race_finsihed(date, current_date):
    """Check if the event date has passed."""
    return date < current_date

passed_races = []

for event_date  in schedule['EventDate']:
    event_date = pd.Timestamp(event_date).date()
    if race_finsihed(event_date, current_date):
        if schedule[schedule['EventDate'] == event_date.strftime('%Y-%m-%d')]['EventFormat'].values[0] != 'testing':
            passed_races.append(schedule[schedule['EventDate'] == event_date.strftime('%Y-%m-%d')])
        else:
            testing = schedule[schedule['EventDate'] == event_date.strftime('%Y-%m-%d')]
    elif not race_finsihed(event_date, current_date):
        event_date = event_date.strftime('%Y-%m-%d')
        upcomming_race = schedule[schedule['EventDate'] == event_date]
        break
        
passed_races
```

```python
laptimes = []
qualifyings = []
results = []

for race in passed_races:
    round_number = race['RoundNumber'].values[0]
    qualifying_session = fastf1.get_session(year, round_number, 'Q')
    qualifying_session.load()
    race_session = fastf1.get_session(year, round_number, 'R')
    race_session.load()


    laptimes_df = race_session.laps
    laptimes_df['RoundNumber'] = round_number
    laptimes.append(laptimes_df)

    qualifying_df = qualifying_session.results
    qualifying_df['RoundNumber'] = round_number
    qualifyings.append(qualifying_df)

    race_results = race_session.results
    race_results['RoundNumber'] = round_number
    results.append(race_results)


combined_laptimes = pd.concat(laptimes, ignore_index=True)
combined_laptimes.to_csv("combined_laptimes.csv", index=False)

# Combine and save qualifying data
combined_qualifyings = pd.concat(qualifyings, ignore_index=True)
combined_qualifyings.to_csv("combined_qualifyings.csv", index=False)

# Combine and save race results
combined_results = pd.concat(results, ignore_index=True)
combined_results.to_csv("combined_results.csv", index=False)
    
    




```
